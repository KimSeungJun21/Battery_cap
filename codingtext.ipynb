```python
def solution(my_string, overwrite_string):
    str1 = list(my_string)
    str2 = list(overwrite_string)
    answer_string = list()
    for i in range(len(str2)):
        answer_string.append(str1[i])
        answer_string.append(str2[i])
    answer = ''.join(answer_string)

    return answer
```

```python
a = solution(my_string, overwrite_string)
print(a)
```

```python
def solution(ineq, eq, n, m):

    if ineq == '<':
        if eq == '=':
            answer = 1 if n <= m else 0
        if eq == '!':
            answer = 1 if n < m else 0
    if ineq == '>':
        if eq == '=':
            answer = 1 if n >= m else 0
        if eq == '!':
            answer = 1 if n > m else 0
    

    return answer
```

```python
a = solution('<', '=', 20,50)
print(a)
```
